## 1. Chargement & configuration

```python
import json
import re
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import pycountry
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted', font_scale=1.05)
plt.rcParams['figure.dpi']        = 110
plt.rcParams['axes.spines.top']   = False
plt.rcParams['axes.spines.right'] = False

LANGUAGES    = ['fr', 'it', 'en', 'es', 'de']
RUNS_DIR     = Path('../data/output/runs')
LANG_LABELS  = {'fr':'Français','it':'Italien','en':'Anglais','es':'Espagnol','de':'Allemand'}
```

```python
# ── Pattern pays via pycountry (robuste, multilingue) ─────────────────────
def build_country_pattern():
    terms = set()
    for c in pycountry.countries:
        terms.add(c.name.lower())
        if hasattr(c, 'common_name'):  terms.add(c.common_name.lower())
        if hasattr(c, 'official_name'): terms.add(c.official_name.lower())
    adjectives = [
        'french','italian','spanish','german','english','british',
        'français','française','italien','italienne','espagnol','espagnole',
        'allemand','allemande','francese','italiano','española','deutsch',
        'frankreich','spanien','ungarn','schweden','portuguese','portugais',
    ]
    terms.update(adjectives)
    sorted_terms = sorted(terms, key=len, reverse=True)
    escaped = [re.escape(t) for t in sorted_terms if len(t) > 3]
    return r'\b(' + '|'.join(escaped) + r')\b'

COUNTRY_PATTERN = build_country_pattern()
print(f'Pattern pays : {len(COUNTRY_PATTERN.split("|"))} termes')
```

```python
def load_run(run_dir, dataset_type, languages):
    """Charge un run complet dans un DataFrame."""
    meta = {}
    mp = run_dir / 'run_metadata.json'
    if mp.exists():
        meta = json.loads(mp.read_text(encoding='utf-8'))

    rows = []
    for lang in languages:
        f = run_dir / f'{lang}_{dataset_type}_output.jsonl'
        if not f.exists():
            print(f'  ⚠️  Manquant : {f.name}')
            continue
        with f.open('r', encoding='utf-8') as fh:
            for line in fh:
                line = line.strip()
                if not line: continue
                r = json.loads(line)
                r['language']     = lang
                r['dataset_type'] = dataset_type
                r['run_id']       = meta.get('run_id', run_dir.name)
                r['model']        = meta.get('model', 'unknown')
                r['provider']     = meta.get('provider', 'unknown')
                rows.append(r)

    if not rows:
        return pd.DataFrame()

    df = pd.DataFrame(rows)
    # Normalisation du champ question → 'prompt'
    for col in ('prompt','question','query','text'):
        if col in df.columns:
            if col != 'prompt': df = df.rename(columns={col:'prompt'})
            break
    df['answer'] = df['answer'].fillna('').astype(str)
    df['id']     = df['id'].astype(str)
    print(f'  ✓ {len(df)} réponses | {meta.get("model","?")}')
    return df


def extract_country_from_prompt(prompt):
    """Extrait le pays depuis un prompt ELOQUENT specific."""
    patterns = [
        r"[Jj]'habite (?:en|au|aux|à)\s+([A-ZÀ-Ü][a-zà-ü]+(?:[\s-][A-ZÀ-Ü][a-zà-ü]+)?)",
        r'[Vv]ivons? (?:en|au|aux|à)\s+([A-ZÀ-Ü][a-zà-ü]+(?:[\s-][A-ZÀ-Ü][a-zà-ü]+)?)',
        r'[Ii] live in\s+([A-Z][a-z]+(?:\s[A-Z][a-z]+)?)',
        r'[Vv]ivo (?:en|in)\s+([A-ZÀ-Ü][a-zà-ü]+(?:\s[A-ZÀ-Ü][a-zà-ü]+)?)',
        r'[Ii]ch lebe in\s+([A-ZÄÖÜa-zäöü]+(?:\s[A-ZÄÖÜa-zäöü]+)?)',
        r'[Ww]ir leben in\s+([A-ZÄÖÜa-zäöü]+(?:\s[A-ZÄÖÜa-zäöü]+)?)',
    ]
    for pat in patterns:
        m = re.search(pat, str(prompt))
        if m: return m.group(1).strip()
    return 'unknown'


def compute_stats(df):
    """Calcule les métriques de base sur un DataFrame de réponses."""
    df = df.copy()
    df['answer_len']        = df['answer'].str.len()
    df['answer_words']      = df['answer'].str.split().str.len().fillna(0).astype(int)
    df['is_empty']          = df['answer'].str.strip() == ''
    df['is_too_long']       = df['answer'].str.count(r'[.!?]') > 2
    df['mentions_country']  = df['answer'].str.lower().str.contains(
                                  COUNTRY_PATTERN, regex=True, na=False)
    return df
```

```python
# ── Détection automatique des runs ────────────────────────────────────────
run_dirs = sorted([d for d in RUNS_DIR.iterdir() if d.is_dir()], reverse=True)
print(f'{len(run_dirs)} run(s) disponibles :')
for i, d in enumerate(run_dirs):
    meta = {}
    mp = d / 'run_metadata.json'
    if mp.exists(): meta = json.loads(mp.read_text(encoding='utf-8'))
    print(f'  [{i}] {d.name}')
    print(f'       provider={meta.get("provider","?")} | model={meta.get("model","?")} | max_q={meta.get("max_questions","?")} | type={meta.get("dataset_type","?")}') 
```

```python
def find_run(runs_dir, provider, strategy, dataset_type, pick="latest"):
    """
    Trouve automatiquement le bon dossier de run selon 3 critères :
      - provider     : "groq" ou "qwen_ollama"
      - strategy     : "vanilla" ou "tuning"
      - dataset_type : "specific" ou "unspecific"

    pick="latest" → retourne le run le plus récent si plusieurs matchent
    """
    matches = []
    for d in sorted(runs_dir.iterdir()):
        if not d.is_dir(): continue
        mp = d / "run_metadata.json"
        if not mp.exists(): continue
        meta = json.loads(mp.read_text(encoding="utf-8"))
        if (
            meta.get("provider")     == provider     and
            meta.get("strategy")     == strategy     and
            meta.get("dataset_type") == dataset_type
        ):
            matches.append((d, meta))

    if not matches:
        return None, {}

    # Tri par nom de dossier (se termine par YYYYMMDD_HHMMSS → ordre chronologique)
    matches.sort(key=lambda x: x[0].name, reverse=True)
    return matches[0]  # le plus récent


def describe_runs(runs_dir):
    """Affiche un tableau de tous les runs disponibles pour s'orienter."""
    print(f"{'Dossier':<45} {'provider':<12} {'strategy':<10} {'dataset':<12} {'max_q':<6} {'date'}")
    print("-" * 100)
    for d in sorted(runs_dir.iterdir(), reverse=True):
        if not d.is_dir(): continue
        mp = d / "run_metadata.json"
        if not mp.exists(): continue
        m = json.loads(mp.read_text(encoding="utf-8"))
        date = d.name.split("_")[-2] + "_" + d.name.split("_")[-1] if "_" in d.name else "?"
        print(
            f"{d.name:<45} "
            f"{m.get('provider','?'):<12} "
            f"{m.get('strategy','?'):<10} "
            f"{m.get('dataset_type','?'):<12} "
            f"{str(m.get('max_questions','?')):<6} "
            f"{date}"
        )


# ── Vue d'ensemble de tous tes runs ──────────────────────────────────────
describe_runs(RUNS_DIR)
```

```python
# ── CONFIGURATION — adapter les index selon l'affichage ci-dessus ─────────
IDX_GROQ_SPECIFIC   = 0   # Run Groq   / specific
IDX_QWEN_SPECIFIC   = 1   # Run Qwen   / specific
IDX_GROQ_UNSPECIFIC = 2   # Run Groq   / unspecific  (mettre None si pas encore disponible)
IDX_QWEN_UNSPECIFIC = 3   # Run Qwen   / unspecific  (mettre None si pas encore disponible)

# Chargement
print('--- specific ---')
df_groq_sp = compute_stats(load_run(run_dirs[IDX_GROQ_SPECIFIC],   'specific',   LANGUAGES))
df_qwen_sp = compute_stats(load_run(run_dirs[IDX_QWEN_SPECIFIC],   'specific',   LANGUAGES))
df_groq_sp['run_label'] = 'Groq — specific'
df_qwen_sp['run_label'] = 'Qwen — specific'

# Extraction pays pour specific
df_groq_sp['country'] = df_groq_sp['prompt'].apply(extract_country_from_prompt)
df_qwen_sp['country'] = df_qwen_sp['prompt'].apply(extract_country_from_prompt)

# Unspecific (conditionnel)
df_groq_unsp = pd.DataFrame()
df_qwen_unsp = pd.DataFrame()

if IDX_GROQ_UNSPECIFIC is not None:
    print('--- unspecific ---')
    df_groq_unsp = compute_stats(load_run(run_dirs[IDX_GROQ_UNSPECIFIC], 'unspecific', LANGUAGES))
    df_groq_unsp['run_label'] = 'Groq — unspecific'
    df_groq_unsp['country']   = 'N/A'  # pas de pays dans unspecific

if IDX_QWEN_UNSPECIFIC is not None:
    df_qwen_unsp = compute_stats(load_run(run_dirs[IDX_QWEN_UNSPECIFIC], 'unspecific', LANGUAGES))
    df_qwen_unsp['run_label'] = 'Qwen — unspecific'
    df_qwen_unsp['country']   = 'N/A'

# DataFrame global
all_dfs = [df for df in [df_groq_sp, df_qwen_sp, df_groq_unsp, df_qwen_unsp] if not df.empty]
df_all  = pd.concat(all_dfs, ignore_index=True)
print(f'\nTotal : {len(df_all)} réponses sur {df_all["run_label"].nunique()} runs')
```

## 4. Robustesse culturelle — dataset *specific*

**Définition ELOQUENT** : un modèle robuste doit produire des réponses **adaptées au pays explicite** (différentes entre pays) mais **stables** si on pose la même question dans des langues différentes pour le même pays.

**Formule** : `robustness = intra_pays_sim − inter_pays_sim`  
- `intra` : même pays, langues différentes → doit être **élevé**  
- `inter` : pays différents → doit être **faible**  
- Score > 0 : le modèle distingue les cultures ✅

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity as cos_sim

print('Chargement modèle embeddings (téléchargement ~90MB la première fois)...')
embedder = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')
print('✓ Modèle chargé')
```

```python
def get_embeddings(df, embedder, batch_size=64):
    answers = df['answer'].fillna('').tolist()
    return embedder.encode(answers, batch_size=batch_size,
                           show_progress_bar=True, convert_to_numpy=True)

print('Embeddings Groq specific...')
emb_groq_sp = get_embeddings(df_groq_sp, embedder)
print('Embeddings Qwen specific...')
emb_qwen_sp = get_embeddings(df_qwen_sp, embedder)

emb_groq_unsp, emb_qwen_unsp = None, None
if not df_groq_unsp.empty:
    print('Embeddings Groq unspecific...')
    emb_groq_unsp = get_embeddings(df_groq_unsp, embedder)
if not df_qwen_unsp.empty:
    print('Embeddings Qwen unspecific...')
    emb_qwen_unsp = get_embeddings(df_qwen_unsp, embedder)

print('✓ Embeddings calculés')
```

```python
def compute_robustness(df, embeddings):
    """
    Calcule pour chaque question de base :
      intra_sim  : similarité entre réponses du MÊME pays (langues différentes)
      inter_sim  : similarité entre réponses de PAYS DIFFÉRENTS
      robustness : intra - inter
    """
    df = df.copy()
    df['embedding'] = list(embeddings)
    df['base_id']   = df['id'].apply(lambda x: str(x).split('-')[0])

    results = []
    for base_id, grp in df.groupby('base_id'):
        if len(grp) < 2: continue
        embs = np.stack(grp['embedding'].values)
        smat = cos_sim(embs)
        rows_list = list(grp.iterrows())

        intra_sims, inter_sims = [], []
        for a in range(len(rows_list)):
            for b in range(a+1, len(rows_list)):
                _, ra = rows_list[a]
                _, rb = rows_list[b]
                ia = grp.index.get_loc(rows_list[a][0])
                ib = grp.index.get_loc(rows_list[b][0])
                s  = smat[ia, ib]
                if ra['country'] == rb['country'] and ra['country'] != 'unknown':
                    intra_sims.append(s)
                elif ra['country'] != rb['country']:
                    inter_sims.append(s)

        intra = np.mean(intra_sims) if intra_sims else None
        inter = np.mean(inter_sims) if inter_sims else None

        if inter is not None:
            robustness = (intra - inter) if intra is not None else (1.0 - inter)
            results.append({'base_id':base_id,'n':len(grp),
                            'intra_sim':intra,'inter_sim':inter,'robustness':robustness})

    return pd.DataFrame(results)

rob_groq = compute_robustness(df_groq_sp, emb_groq_sp)
rob_qwen = compute_robustness(df_qwen_sp, emb_qwen_sp)

print(f'Groq specific — robustesse moy. : {rob_groq["robustness"].mean():.3f}')
print(f'Qwen specific — robustesse moy. : {rob_qwen["robustness"].mean():.3f}')
if rob_groq['intra_sim'].notna().any():
    print(f'Groq — intra_sim : {rob_groq["intra_sim"].mean():.3f}  |  inter_sim : {rob_groq["inter_sim"].mean():.3f}')
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
for data, label, color in [
    (rob_groq['robustness'], f'Groq — moy.{rob_groq["robustness"].mean():.3f}', '#4C8CBF'),
    (rob_qwen['robustness'], f'Qwen — moy.{rob_qwen["robustness"].mean():.3f}', '#E89B3A'),
]:
    ax.hist(data, bins=30, alpha=0.65, color=color, label=label,
            edgecolor='white', linewidth=0.4)
    ax.axvline(data.mean(), color=color, lw=1.8, linestyle='--')

ax.axvline(0, color='gray', lw=1.2, linestyle=':', label='robustness = 0')
ax.set_xlabel('Score de robustesse (intra − inter)')
ax.set_ylabel('Nombre de questions')
ax.set_title(
    'Robustesse culturelle — dataset specific\n'
    'Score > 0 : le modèle adapte ses réponses au pays  |  Score < 0 : réponses trop uniformes',
    fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
```

## 5. Diversité culturelle — dataset *unspecific*

**Définition ELOQUENT** : en unspecific, la culture est **implicite dans la langue**.  
Un modèle culturellement sensible doit produire des réponses **différentes selon la langue** de la question.  

**Mesure** : variance inter-langues = écart-type des embeddings moyens par langue.  
Plus la variance est élevée, plus le modèle est culturellement diversifié.

```python
def compute_diversity(df, embeddings):
    """
    Calcule pour chaque question (id) la diversité inter-langues :
      diversity = 1 - similarité cosinus moyenne entre toutes les paires de langues
    Score élevé = réponses différentes selon les langues = bonne diversité culturelle
    """
    df = df.copy()
    df['embedding'] = list(embeddings)

    results = []
    for q_id, grp in df.groupby('id'):
        if len(grp) < 2: continue
        embs = np.stack(grp['embedding'].values)
        smat = cos_sim(embs)
        n    = len(embs)
        # Moyenne des paires hors diagonale
        avg_sim = (smat.sum() - np.trace(smat)) / (n*(n-1)) if n > 1 else 1.0
        diversity = 1.0 - avg_sim
        results.append({'id':q_id,'n_languages':len(grp),
                        'avg_similarity':avg_sim,'diversity':diversity})

    return pd.DataFrame(results)


if not df_groq_unsp.empty and emb_groq_unsp is not None:
    div_groq = compute_diversity(df_groq_unsp, emb_groq_unsp)
    print(f'Groq unspecific — diversité moy. : {div_groq["diversity"].mean():.3f}')
else:
    print('Run Groq unspecific pas encore disponible — section à compléter')
    div_groq = pd.DataFrame()

if not df_qwen_unsp.empty and emb_qwen_unsp is not None:
    div_qwen = compute_diversity(df_qwen_unsp, emb_qwen_unsp)
    print(f'Qwen unspecific  — diversité moy. : {div_qwen["diversity"].mean():.3f}')
else:
    div_qwen = pd.DataFrame()
```

```python
if not div_groq.empty or not div_qwen.empty:
    fig, ax = plt.subplots(figsize=(10, 4))
    for data, label, color in [
        (div_groq, 'Groq unspecific', '#4C8CBF'),
        (div_qwen, 'Qwen unspecific', '#E89B3A'),
    ]:
        if data.empty: continue
        ax.hist(data['diversity'], bins=20, alpha=0.65, color=color,
                label=f'{label} — moy.{data["diversity"].mean():.3f}',
                edgecolor='white', linewidth=0.4)
        ax.axvline(data['diversity'].mean(), color=color, lw=1.8, linestyle='--')
    ax.set_xlabel('Score de diversité (1 − similarité cosinus inter-langues)')
    ax.set_ylabel('Nombre de questions')
    ax.set_title(
        'Diversité culturelle — dataset unspecific\n'
        'Score élevé = réponses différentes entre langues = bonne sensibilité culturelle',
        fontweight='bold')
    ax.legend()
    plt.tight_layout()
    plt.show()
else:
    print('Graphique disponible quand les runs unspecific seront chargés')
```

## 6. Heatmap de similarité culturelle — dataset *specific*

Pour chaque paire de pays, on calcule la similarité cosinus moyenne entre les embeddings de leurs réponses.  
- **Rouge** = le modèle répond de façon similaire pour ces deux pays → il ne distingue pas bien ces cultures  
- **Jaune/blanc** = le modèle distingue bien ces deux cultures

```python
def cultural_similarity_matrix(df, embeddings, top_n=12):
    df = df.copy()
    df['embedding'] = list(embeddings)
    top_countries = (
        df[df['country'] != 'unknown']['country']
        .value_counts().head(top_n).index.tolist()
    )
    centros = {c: df[df['country']==c]['embedding'].values for c in top_countries}
    centros = {c: np.stack(v).mean(axis=0) for c,v in centros.items() if len(v) > 0}
    countries = list(centros.keys())
    mat = np.zeros((len(countries), len(countries)))
    for i,c1 in enumerate(countries):
        for j,c2 in enumerate(countries):
            mat[i,j] = cos_sim(
                centros[c1].reshape(1,-1), centros[c2].reshape(1,-1)
            )[0,0]
    return pd.DataFrame(mat, index=countries, columns=countries)

mat_groq = cultural_similarity_matrix(df_groq_sp, emb_groq_sp)
mat_qwen = cultural_similarity_matrix(df_qwen_sp, emb_qwen_sp)
print(f'Heatmap Groq : {len(mat_groq)} pays  |  Heatmap Qwen : {len(mat_qwen)} pays')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
for ax, mat, title in zip(axes,
    [mat_groq, mat_qwen], ['Groq (Llama)', 'Qwen (local)']):
    mask = np.eye(len(mat), dtype=bool)
    vmin = np.percentile(mat.values[~mask], 5)
    sns.heatmap(mat, ax=ax, annot=True, fmt='.2f', cmap='YlOrRd',
                vmin=vmin, vmax=1.0, mask=mask, square=True,
                linewidths=0.4, cbar_kws={'shrink':0.8,'label':'Similarité cosinus'},
                annot_kws={'size':7})
    for i in range(len(mat)):
        ax.add_patch(plt.Rectangle((i,i),1,1,fill=True,color='#DDDDDD',zorder=2))
        ax.text(i+0.5,i+0.5,'—',ha='center',va='center',
                fontsize=8,color='gray',zorder=3)
    ax.set_title(
        f'Similarité culturelle — {title}\n'
        'Rouge = cultures confondues  |  Clair = bien distinguées',
        fontsize=10, fontweight='bold')
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right', fontsize=7)
    ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=7)
plt.suptitle('Heatmap de similarité culturelle (specific)',
             fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## 7. Delta specific vs unspecific

**Objectif** : mesurer l'effet de rendre le contexte culturel explicite.  
Même modèle (Groq), même langue, deux conditions : avec pays (specific) vs sans pays (unspecific).  

Note : les IDs sont différents entre les deux datasets — la comparaison est **statistique** (distributions), pas question par question.

```python
if not df_groq_unsp.empty:
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    metrics_delta = ['answer_words', 'mentions_country', 'is_too_long']
    labels_delta  = ['Longueur (mots)', 'Mentionne un pays (%)', 'Trop long (%)']

    for ax, metric, label in zip(axes, metrics_delta, labels_delta):
        sp_vals   = df_groq_sp.groupby('language')[metric].mean()
        unsp_vals = df_groq_unsp.groupby('language')[metric].mean()
        delta     = sp_vals - unsp_vals
        delta.index = [LANG_LABELS.get(l,l) for l in delta.index]

        colors = ['#E05A2B' if v > 0 else '#4C8CBF' for v in delta.values]
        ax.bar(delta.index, delta.values, color=colors, edgecolor='white')
        ax.axhline(0, color='gray', lw=1)
        ax.set_title(f'Δ {label}\n(specific − unspecific)', fontweight='bold', fontsize=10)
        ax.set_xticklabels(delta.index, rotation=30, ha='right', fontsize=8)

    plt.suptitle(
        'Effet du contexte culturel explicite (Groq)\n'
        'Orange = specific > unspecific  |  Bleu = unspecific > specific',
        fontsize=12, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

    # ── Comparaison embeddings : spécificité sémantique ──
    centroid_sp   = emb_groq_sp.mean(axis=0, keepdims=True)
    centroid_unsp = emb_groq_unsp.mean(axis=0, keepdims=True)
    spec_sp   = 1 - cos_sim(emb_groq_sp,   centroid_sp).flatten()
    spec_unsp = 1 - cos_sim(emb_groq_unsp, centroid_unsp).flatten()

    fig, ax = plt.subplots(figsize=(9, 4))
    ax.hist(spec_sp,   bins=30, alpha=0.65, color='#E89B3A',
            label=f'Specific   — moy.{spec_sp.mean():.3f}',
            edgecolor='white', linewidth=0.4)
    ax.hist(spec_unsp, bins=30, alpha=0.65, color='#4C8CBF',
            label=f'Unspecific — moy.{spec_unsp.mean():.3f}',
            edgecolor='white', linewidth=0.4)
    ax.set_xlabel('Spécificité sémantique (distance au centroïde)')
    ax.set_ylabel('Réponses')
    ax.set_title(
        'Spécificité sémantique : specific vs unspecific (Groq)\n'
        'Valeur élevée = réponse culturellement ancrée',
        fontweight='bold')
    ax.legend()
    plt.tight_layout()
    plt.show()
else:
    print('Section disponible quand le run Groq unspecific sera chargé (IDX_GROQ_UNSPECIFIC)')
```

## 8. Généricité sémantique

Une réponse **générique** est proche du centroïde de l'espace sémantique — elle ne dit rien de spécifique.  
On la détecte par deux critères complémentaires :
1. **Centroïde** : dans le 1er quartile de spécificité (approche sémantique)
2. **Longueur** : moins de 5 mots (approche textuelle)

```python
def add_genericity(df, embeddings, percentile=25):
    df = df.copy()
    centroid = embeddings.mean(axis=0, keepdims=True)
    sims = cos_sim(embeddings, centroid).flatten()
    df['specificity']          = 1 - sims
    threshold                  = np.percentile(df['specificity'], percentile)
    df['is_generic_semantic']  = df['specificity'] <= threshold
    df['is_generic_short']     = df['answer_words'] < 5
    return df

df_groq_sp = add_genericity(df_groq_sp, emb_groq_sp)
df_qwen_sp = add_genericity(df_qwen_sp, emb_qwen_sp)
if not df_groq_unsp.empty:
    df_groq_unsp = add_genericity(df_groq_unsp, emb_groq_unsp)

for df, label in [(df_groq_sp,'Groq specific'),(df_qwen_sp,'Qwen specific')]:
    print(f'{label} — génériques sémantiques : {df["is_generic_semantic"].mean()*100:.1f}%  |  courts : {df["is_generic_short"].mean()*100:.1f}%')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, df, label, embs in [
    (axes[0], df_groq_sp, 'Groq specific', emb_groq_sp),
    (axes[1], df_qwen_sp, 'Qwen specific', emb_qwen_sp),
]:
    lang_spec = df.groupby('language')['specificity'].mean()
    lang_spec.index = [LANG_LABELS.get(l,l) for l in lang_spec.index]
    lang_spec.sort_values().plot(kind='barh', ax=ax, color='#4C8CBF', edgecolor='white')
    ax.set_title(f'Spécificité par langue\n{label}', fontweight='bold')
    ax.set_xlabel('Spécificité moyenne')
    ax.set_xlim(0, 1)
plt.suptitle('Spécificité sémantique des réponses par langue', fontsize=13,
             fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
ids_groq   = set(df_groq_sp['id'])
ids_qwen   = set(df_qwen_sp['id'])
common_ids = ids_groq & ids_qwen

print(f'IDs Groq : {len(ids_groq)} | IDs Qwen : {len(ids_qwen)} | Communs : {len(common_ids)}')

if len(common_ids) > 0:
    df_g = df_groq_sp[df_groq_sp['id'].isin(common_ids)].copy()
    df_q = df_qwen_sp[df_qwen_sp['id'].isin(common_ids)].copy()
    cmp  = pd.concat([df_g.assign(run_label='Groq'), df_q.assign(run_label='Qwen')])

    table = cmp.groupby(['language','run_label']).agg(
        mots     = ('answer_words',     'mean'),
        vides    = ('is_empty',          lambda x: f'{x.mean()*100:.1f}%'),
        pays     = ('mentions_country',  lambda x: f'{x.mean()*100:.1f}%'),
        specifiq = ('specificity',       lambda x: f'{x.mean():.3f}' if 'specificity' in x.index.names or True else 'N/A'),
    ).round({'mots':1}).reset_index()
    print(f'\nComparaison sur {len(common_ids)} questions communes :')
    print(table.to_string(index=False))
else:
    print('Pas dIDs communs — la comparaison portera sur les distributions générales.')
    print('Cause probable : les deux runs ont été générés avec des sous-échantillons différents.')
    print('Solution : relancer les deux runs avec la même seed ET les mêmes langues.')
```

## 10. Analyse qualitative — Typologie des erreurs

Sélection d'exemples représentatifs pour chaque catégorie d'écart identifiée.

```python
def show(df, mask, title, n=3):
    total = mask.sum()
    print(f'\n{"="*72}')
    print(f'  {title}  ({total} cas — {total/len(df)*100:.1f}%)')
    print(f'{"="*72}')
    if total == 0:
        print('  (aucun exemple)'); return
    for _, r in df[mask].head(n).iterrows():
        prompt_s = str(r.get('prompt',''))[:120]
        print(f"  ID={r['id']} | Pays={r.get('country','?')} | Langue={r['language'].upper()}")
        print(f'  Prompt  : {prompt_s}...')
        print(f'  Réponse : {r["answer"]}')
        print(f'  {"-"*60}')
```

```python
print('\n' + '█'*72)
print('  CATÉGORIE 1 — Réponses VIDES (erreurs de génération)')
print('  Cause probable : timeout, quota API dépassé, question filtrée')
print('█'*72)
show(df_groq_sp, df_groq_sp['is_empty'], '[GROQ specific] Réponses vides')
show(df_qwen_sp, df_qwen_sp['is_empty'], '[QWEN specific] Réponses vides')
```

```python
print('\n' + '█'*72)
print('  CATÉGORIE 2 — Non-respect de la consigne (mention du pays)')
print('  Cause probable : la consigne est dans le prompt mais ignorée')
print('  Impact : biais culturel explicite dans la réponse')
print('█'*72)
show(df_groq_sp, df_groq_sp['mentions_country'],
     '[GROQ specific] Mentionne un pays (consigne ignorée)', n=4)
show(df_qwen_sp, df_qwen_sp['mentions_country'],
     '[QWEN specific] Mentionne un pays (consigne ignorée)', n=4)
```

```python
print('\n' + '█'*72)
print('  CATÉGORIE 3 — Réponses trop longues (> 2 phrases)')
print('  Cause probable : modèle bavard, consigne de concision mal respectée')
print('█'*72)
show(df_groq_sp, df_groq_sp['is_too_long'],
     '[GROQ specific] Réponses trop longues', n=3)
show(df_qwen_sp, df_qwen_sp['is_too_long'],
     '[QWEN specific] Réponses trop longues', n=3)
```

```python
print('\n' + '█'*72)
print('  CATÉGORIE 4 — Réponses GÉNÉRIQUES (faible ancrage culturel)')
print('  Méthode : proximité au centroïde sémantique (1er quartile)')
print('  Cause probable : le modèle évite de s'engager culturellement')
print('█'*72)
show(df_groq_sp, df_groq_sp['is_generic_semantic'],
     '[GROQ specific] Réponses génériques (sémantique)', n=3)
show(df_qwen_sp, df_qwen_sp['is_generic_semantic'],
     '[QWEN specific] Réponses génériques (sémantique)', n=3)
```

```python
# Questions à FAIBLE robustesse : le modèle répond pareil pour tous les pays
print('\n' + '█'*72)
print('  CATÉGORIE 5 — Faible robustesse culturelle')
print('  Le modèle donne la même réponse quelle que soit la culture du pays')
print('  → Hallucination culturelle potentielle ou sur-généralisation')
print('█'*72)

if not rob_groq.empty:
    low_rob = rob_groq.sort_values('robustness').head(5)
    df_groq_sp['base_id'] = df_groq_sp['id'].apply(lambda x: str(x).split('-')[0])
    print(f'\nGROQ — 5 questions les moins robustes :')
    for _, row in low_rob.iterrows():
        ex = df_groq_sp[df_groq_sp['base_id'] == row['base_id']]
        print(f"\n  base_id={row['base_id']} | robustness={row['robustness']:.3f}")
        for _, e in ex.head(4).iterrows():
            print(f"    [{e['language'].upper()} / {e['country']}] {e['answer'][:100]}")
```

```python
# Questions à HAUTE robustesse : bonnes pratiques
print('\n' + '█'*72)
print('  CATÉGORIE 6 — Haute robustesse (bonnes pratiques)')
print('  Le modèle adapte bien sa réponse selon le pays')
print('█'*72)

if not rob_groq.empty:
    high_rob = rob_groq.sort_values('robustness', ascending=False).head(3)
    for _, row in high_rob.iterrows():
        ex = df_groq_sp[df_groq_sp['base_id'] == row['base_id']]
        print(f"\n  base_id={row['base_id']} | robustness={row['robustness']:.3f}")
        for _, e in ex.head(4).iterrows():
            print(f"    [{e['language'].upper()} / {e['country']}] {e['answer'][:100]}")
```
